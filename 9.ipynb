```python
lam = 0.25
mu = 0.3
n = 3
T = 100000
```

```python
import random

random.seed(42)


def next_arrival():
    return random.expovariate(lam)


def service_time():
    return random.expovariate(mu)
```

```python
queue = []
servers = [0] * n
t = 0

total_arrivals = 0
total_wait = 0.0
total_system = 0.0
busy_time = [0.0] * n

arrival_time = next_arrival()

while t < T:
    next_service = min([s for s in servers if s > t], default=float("inf"))

    if arrival_time < next_service:
        t = arrival_time
        total_arrivals += 1

        free_servers = [i for i in range(n) if servers[i] <= t]

        if free_servers:
            i = free_servers[0]
            st = service_time()

            servers[i] = t + st
            busy_time[i] += st

            total_system += st
        else:
            queue.append(t)

        arrival_time = t + next_arrival()
    else:
        t = next_service
        for i in range(n):
            if servers[i] == t:
                if queue:
                    arrival = queue.pop(0)
                    wait = t - arrival
                    st = service_time()
                    servers[i] = t + st

                    total_wait += wait
                    total_system += wait + st
                    busy_time[i] += st
                else:
                    servers[i] = 0
```

```python
rho = sum(busy_time) / T
T_wait = total_wait / total_arrivals
T_sys = total_system / total_arrivals

print("ρ =", rho)
print("Tож =", T_wait)
print("Tсис =", T_sys)
```

$$\lambda = \frac{5}{20} = \frac{1}{4} \text{ чел/мин}$$

$$\mu = \frac{3}{10} \text{ чел/мин}$$

$$n = 3$$

$$\rho = \frac{\lambda}{\mu} = \frac{5}{6} \approx 0.8(3)$$

$$\psi = \frac{\rho}{n} = \frac{5}{18}$$

$$P_0 = \left[ \sum_{k=0}^n \frac{n^k}{k!} \psi^k + \frac{n^n}{n!} \frac{\psi^{n+1} \left( 1 - \psi^m \right)}{1 - \psi} \right]^{-1}$$

$$P_0 = \left[ \sum_{k=0}^3 \frac{3^k}{k!} \left(\frac{5}{18}\right)^k + \frac{3^3}{3!} \frac{\frac{5}{18}^{3+1} \left(1 - \psi^{\infty} \right)}{1 - \frac{5}{18}} \right]^{-1}$$

$$P_0 = \frac{156}{361}$$

$$P_\text{сис} = 1 - \frac{n^n}{n!} \psi^{n + m} P_0 = 1 - \frac{3^3}{3!} \psi^{n + \infty} P_0 = 1$$

$$\overline{N}_\text{об} = \rho P_\text{сис} = \frac{5}{6}$$

$$\overline{N}_\text{ож} = \frac{n^n}{n!} \psi^{n + 1} \frac{1 - (m + 1) \psi^m + m \psi^{m + 1}}{\left(1 - \psi \right)^2}P_0$$

$$\overline{N}_\text{ож} = \frac{3^3}{3!} \left(\frac{5}{18} \right)^{3 + 1} \frac{1 - (\infty + 1) \psi^\infty + \infty \cdot \psi^{\infty + 1}}{\left(1 - \frac{5}{18} \right)^2}P_0$$

$$\overline{N}_\text{ож} = \frac{625}{25158}$$

$$\overline{T}_\text{ож} = \frac{\overline{N}_\text{ож}}{\lambda P_\text{сис}} = \frac{\frac{625}{25158}}{\frac{1}{4} \cdot 1} = \frac{1250}{14079} \approx 0.0887847148$$

$$\overline{N}_\text{сис} = \overline{N}_\text{ож} + \overline{N}_\text{об} = \frac{625}{25158} + \frac{5}{6} = \frac{4015}{4693}$$

$$\overline{T}_\text{сис} = \frac{\overline{N}_\text{сис}}{\lambda P_\text{сис}} = \frac{\frac{4015}{4693}}{\frac{1}{4} \cdot 1} = \frac{16060}{4693} \approx 3.4221180482$$
